```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/adity/Downloads/Copy of output.csv", encoding='latin1')
print(df.head(5))
```

```python
#nan values
print(df.isnull().sum())
len(df)
```

```python
df = df.drop(columns=['description'])
df = df.dropna(subset=['selling_price'])
df['average_rating'] = df['average_rating'].fillna(0)
df['brand'] = df['brand'].fillna('Unknown')
df['seller'] = df['seller'].fillna('Unknown')
df['actual_price'] = df['actual_price'].fillna(df['selling_price'])
```

```python
# Removed nan values
print(df.isnull().sum())
len(df)
```

```python
df['selling_price'] = df['selling_price'].replace('[₹,]', '', regex=True).astype(float)
df['actual_price'] = df['actual_price'].replace('[₹,]', '', regex=True).astype(float)
df[['selling_price','actual_price']].dtypes
```

```python
df['discount_amount'] = df['actual_price'] - df['selling_price']
df['discount_percent'] = (df['discount_amount'] / df['actual_price']) * 100
print(df[['discount_amount', 'discount_percent']].head(5))
```

```python
total_revenue = df['selling_price'].sum()
print("Total Revenue:", total_revenue)
```
