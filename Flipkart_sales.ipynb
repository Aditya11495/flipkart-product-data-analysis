```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/adity/Downloads/Copy of output.csv", encoding='latin1')
print(df.head(5))
```

```python
#nan values
print(df.isnull().sum())
len(df)
```

```python
df = df.drop(columns=['description'])
df = df.dropna(subset=['selling_price'])
df['average_rating'] = df['average_rating'].fillna(0)
df['brand'] = df['brand'].fillna('Unknown')
df['seller'] = df['seller'].fillna('Unknown')
df['actual_price'] = df['actual_price'].fillna(df['selling_price'])
```

```python
# Removed nan values
print(df.isnull().sum())
len(df)
```

```python
df['selling_price'] = df['selling_price'].replace('[₹,]', '', regex=True).astype(float)
df['actual_price'] = df['actual_price'].replace('[₹,]', '', regex=True).astype(float)
df[['selling_price','actual_price']].dtypes
```

```python
df['discount_amount'] = df['actual_price'] - df['selling_price']
df['discount_percent'] = (df['discount_amount'] / df['actual_price']) * 100
print(df[['discount_amount', 'discount_percent']].head(5))
```

```python
total_revenue = df['selling_price'].sum()
print("Total Revenue:", total_revenue)
```

```python

category_revenue = df.groupby('category')['selling_price'].sum().sort_values(ascending=False)

# Plot
plt.figure()
category_revenue.plot(kind='bar')
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.title("Revenue by Category")
plt.xticks(rotation=45)
plt.show()
```

```python
category_discount = df.groupby('category')['discount_percent'].mean().sort_values(ascending=False)

plt.figure()
category_discount.plot(kind='bar')
plt.xlabel("Category")
plt.ylabel("Average Discount %")
plt.title("Average Discount Percentage by Category")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure()
plt.hist(df['average_rating'], bins=10)
plt.xlabel("Average Rating")
plt.ylabel("Count")
plt.title("Distribution of Product Ratings")
plt.show()
```

```python
plt.figure()
plt.scatter(df['discount_percent'], df['selling_price'])
plt.xlabel("Discount %")
plt.ylabel("Selling Price")
plt.title("Discount % vs Selling Price")
plt.show()
```
